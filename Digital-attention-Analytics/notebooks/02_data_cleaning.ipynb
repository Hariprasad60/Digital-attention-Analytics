```python
import pandas as pd

# Load the dataset from your raw folder
df = pd.read_csv('../data/raw/smartphone-usage-data.csv')

# Drop duplicates
df = df.drop_duplicates()

# Fill missing values: median for numbers, mode for text
for col in df.columns:
    if df[col].dtype in ['int64', 'float64']:
        df[col] = df[col].fillna(df[col].median())
    else:
        df[col] = df[col].fillna(df[col].mode()[0])

# Save the cleaned dataset to the processed folder
cleaned_path = '../data/processed/cleaned_smartphone_usage.csv'
df.to_csv(cleaned_path, index=False)

print("Data cleaning complete!")
print(f"Cleaned dataset saved to: {cleaned_path}")
print(f"Final shape: {df.shape}")
```
