```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set visualization style
sns.set_theme(style="whitegrid")

# 1. Load the cleaned data
df = pd.read_csv('../data/processed/cleaned_smartphone_usage.csv')

# 2. FEATURE ENGINEERING
# Combine specific minutes into a total tracked activity feature
activity_cols = ['SocialMedia_Min', 'Gaming_Min', 'Study_Min', 'Calls_Min']
# Only use columns that actually exist in the dataset
available_cols = [col for col in activity_cols if col in df.columns]

if available_cols:
    df['Total_Tracked_Activity_Hours'] = df[available_cols].sum(axis=1) / 60

# Save the newly engineered dataset
engineered_path = '../data/processed/engineered_smartphone_usage.csv'
df.to_csv(engineered_path, index=False)
print("Feature Engineering complete. File saved!\n")

# 3. EXPLORATORY DATA ANALYSIS (EDA)
# Plot 1: Screen Time Distribution
if 'Daily_ScreenTime_Hours' in df.columns:
    plt.figure(figsize=(8, 5))
    sns.histplot(df['Daily_ScreenTime_Hours'], kde=True, color='blue', bins=30)
    plt.title('Distribution of Daily Screen Time')
    plt.xlabel('Hours')
    plt.ylabel('Frequency')
    plt.show()

# Plot 2: Correlation Heatmap
plt.figure(figsize=(10, 8))
# Select only numeric columns for correlation
numeric_df = df.select_dtypes(include=['float64', 'int64'])
corr = numeric_df.corr()
sns.heatmap(corr, annot=False, cmap='coolwarm', fmt=".2f")
plt.title('Feature Correlation Matrix')
plt.show()
```
